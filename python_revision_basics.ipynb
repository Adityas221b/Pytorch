```python
##MAP AND FILTER

x=[1,2,33,34,12,122,133,13331,1332,1332,55332,11]
mp=map(lambda i:i+2,x)
print(list(mp))
```

```python
##MAP AND FILTER

x=[1,2,33,34,12,122,133,13331,1332,1332,55332,11]
mp=filter(lambda i:i%2==0,x)
print(list(mp))
```
