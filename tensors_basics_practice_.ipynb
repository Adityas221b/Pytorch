##RESHAPING STACKING SQUEZZING AND UNSQUEZZING TENSORS
*RESHAPING -SHAPES AN INPUT TENSORS ON YOP OF CERTAIN SHAPE BUT KEEP THE SAME MEMORY AS THE ORIGNAL TENSOR
*STACKING -COMBINNING MULTIPLE TENSORS  ON TOP OF EACH OTHER(VSTACK)
*SQUEEZE -REMOVES '1 DIMENTIONS FROM A TENSOR
*UNSQUEEZE -ADD ONE DIMENTION TO THE TARGET HERE
*PERMUTE-RETURN A VIEW OF THE INPUT WITH DIMENTIONS PERMUTED (SWAPPED) IN A CERTAIN WAY

```python
import torch
x=torch.arange(1.,13.)
x,x.shape
```

```python
y=torch.arange(1.,13.)
y,y.shape
```

```python
y=y.reshape(6,2)
```

```python
# #reproducability (trying to make random out of random in here )
# start with random number -> tensor operations ->update random number here ->
# try and make better production of data ->again and again


# to reduce the randomness in NN and pytorch comes the concept of random seed
# seed does is flovour randomness
```

```python
import torch
a=torch.rand(3,4)
b=torch.rand(3,4)
print(a)
print(b)
print(a==b)
```

```python
import torch
RANDOM=42
torch.manual_seed(RANDOM)
c=torch.rand(3,4)
torch.manual_seed(RANDOM)
d=torch.rand(3,4)

print(c)
print(d)
print(c==d)
```
